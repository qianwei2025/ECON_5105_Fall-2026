## Option 1 — Marketing A/B testing

**Question:** Did showing an **advertisement** (versus a **PSA**) raise the **conversion** probability? Is the difference **statistically** and **economically** meaningful?

**Data:** [Marketing A/B Testing](https://www.kaggle.com/datasets/faviovaz/marketing-ab-testing/data) (Favio Vázquez).

| Column | Meaning |
|--------|--------|
| `user id` | Unique user |
| `test group` | `ad` = saw the advertisement; `psa` = saw a public service announcement |
| `converted` | Bought the product (`True`/`False`) |
| `total ads` | Number of ads seen |
| `most ads day` / `most ads hour` | When the user saw the most ads |

**Methods (examples):** Two-sample test of proportions (or $\chi^2$ / $z$-test); **confidence interval** for the lift; check **balance** if you use covariates; **power** or **minimum detectable effect** given the realized sample; optional **heterogeneity** by day/hour (watch **multiple testing**). Discuss practical significance (lift in conversion, not only $p$-values).

**Watch-outs:** The ad arm is much larger than the PSA arm. Do not treat `total ads` as randomly assigned. If you slice the data many ways, say how you control false positives.

## Option 2 — Mobile game A/B testing (Cookie Cats)

**Question:** Should Cookie Cats keep the first in-game **gate at level 30** or move it to **level 40**? What happens to **1-day retention**, **7-day retention**, and **game rounds played**?

This is a step beyond a single conversion-rate test: two binary retention outcomes, a highly skewed count outcome, and a product decision that can go in **either** direction.

**Data:** [Mobile Games A/B Testing — Cookie Cats](https://www.kaggle.com/datasets/mursideyarkin/mobile-games-ab-testing-cookie-cats) (also on [Datacamp/Tactile Entertainment](https://www.kaggle.com/datasets/yufengsui/mobile-games-ab-testing)). About 90,000 players randomly assigned at install.

| Column | Meaning |
|--------|--------|
| `userid` | Unique player |
| `version` | `gate_30` (control) or `gate_40` (treatment) |
| `sum_gamerounds` | Rounds played in the first 14 days |
| `retention_1` | Came back 1 day after install |
| `retention_7` | Came back 7 days after install |

**Alternative Kaggle dataset (multi-arm test):** [Fast Food Marketing Campaign A/B Test](https://www.kaggle.com/datasets/chebotinaa/fast-food-marketing-campaign-ab-test) — three promotions, store–week **sales**. If you choose this instead, compare the three arms (ANOVA / pairwise tests with a multiple-comparison correction) and account for **repeated weeks within stores**.

**Methods (examples, Cookie Cats):** Check **sample-ratio mismatch** (assignment 50/50?). Test **retention_1** and **retention_7** with two-sample proportion tests and CIs. For `sum_gamerounds`, do **not** rely on a naive $t$-test alone—the distribution is skewed with extreme outliers; use a **bootstrap** CI, a rank test, and/or a trimmed/winsorized mean. Pre-specify a **primary metric** (usually 7-day retention) and treat the others as secondary; correct for **multiple testing** if you claim several “wins.” Optional: **power** / MDE, and a recommendation (ship gate_40 or keep gate_30).

**Watch-outs:** Players who never reach level 30/40 are still in the data (intent-to-treat). A significant 1-day result with a **null or negative** 7-day result is still a finding—do not hide it. Extreme `sum_gamerounds` values can dominate means.

## Option 3 — House prices and linear regression

**Question:** Which housing and neighborhood characteristics predict **house values**, and how large are the **partial effects** after controls?

**Data (pick one):**
- **Recommended:** [California Housing](https://scikit-learn.org/stable/datasets/real_world.html#california-housing-dataset) — `sklearn.datasets.fetch_california_housing`. Median house value by census block group; regressors include median income, house age, rooms, population, and location.
- **Classic alternative (used in prior years):** [Boston Housing](https://www.cs.toronto.edu/~delve/data/boston/bostonDetail.html). If you use Boston, **do not** treat the `B` variable as a neutral control; discuss what it measures and why many libraries no longer ship this dataset.

**Methods (examples):** Start with a **simple** regression (e.g. value on income or rooms), then a **multiple** regression. Report heteroskedasticity-robust SEs; residual plots; a small **specification** table (add controls in columns). Optional train/test split if you emphasize **prediction**, but coefficient interpretation should still be the center of the paper.

**Watch-outs:** Coefficients are **ceteris paribus associations**, not causal effects of “building more rooms.” Discuss multicollinearity (rooms vs. bedrooms, location vs. income) and outliers.

## Option 4 — Panel data (entity and time fixed effects)

**Question:** After you account for **stable differences across units** and **common shocks over time**, what association remains between a policy or economic regressor and the outcome?

**Data (pick one):**
- **U.S. traffic fatalities and beer taxes (state–year):** [AER `Fatalities`](https://vincentarelbundock.github.io/Rdatasets/csv/AER/Fatalities.csv) — classic panel used to study alcohol policy and highway deaths. Codebook: [Rdatasets Fatalities](https://vincentarelbundock.github.io/Rdatasets/doc/AER/Fatalities.html).
- **Wooldridge `wagepan`:** `wooldridge.data('wagepan')` — NLSY worker–year panel; wages, experience, union, marriage, education (education is time-invariant for most workers—use that as a teaching point about what FE can and cannot identify).

**Methods (examples):** Pooled OLS, then entity FE, then two-way FE. Cluster SEs at the **entity** level. Compare coefficients across specifications in one table. Optional: `linearmodels.panel.PanelOLS`.

**Watch-outs:** FE does not fix reverse causality or time-varying confounders. Time-invariant regressors drop out of entity FE. Do not call two-way FE a causal design unless you defend parallel-trends / conditional exogeneity in this application.

## Suggested paper and slide outline

Use this as a checklist, not a rigid template.

1. **Question and hypothesis.** One sentence for the estimand; a testable $H_0$ if you are doing inference.
2. **Data.** Source, unit of observation, sample restrictions, missing values, summary table.
3. **Methods.** The model equation, why OLS/FE/the test is appropriate, and the uncertainty you will report (SEs, CIs, $p$-values, power).
4. **Results.** A small number of clear tables/figures; interpret magnitudes, not only stars.
5. **Robustness.** At least one extra specification (more controls, logs, clustered SEs, alternative sample).
6. **Limitations.** What the design cannot answer.
7. **Conclusion.** What a decision-maker should take away.

**Software:** Python 3 with NumPy, Pandas, SciPy, Statsmodels, matplotlib/seaborn; `scikit-learn` and `linearmodels` where useful (see the syllabus).

**Academic integrity:** Independent write-up and code. You may discuss ideas with classmates. Cite the data source and any code or papers you rely on.
